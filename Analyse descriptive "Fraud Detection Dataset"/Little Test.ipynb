> ## `Hypothèses du test Little`
> Hypothèse nulle (H₀)  : Les données manquantes sont MCAR.
>
> Hypothèse alternative (H₁): Les données ne sont pas MCAR (elles sont MAR ou MNAR).

```python
# Source - https://stackoverflow.com/a

import numpy as np
import pandas as pd
from scipy.stats import chi2

def little_mcar_test(data, alpha=0.05):
    """
    Performs Little's MCAR (Missing Completely At Random) test on a dataset with missing values.
    """
    data = pd.DataFrame(data)
    data.columns = ['x' + str(i) for i in range(data.shape[1])]
    data['missing'] = np.sum(data.isnull(), axis=1)
    n = data.shape[0]
    k = data.shape[1] - 1
    df = k * (k - 1) / 2
    chi2_crit = chi2.ppf(1 - alpha, df)
    chi2_val = ((n - 1 - (k - 1) / 2) ** 2) / (k - 1) / ((n - k) * np.mean(data['missing']))
    p_val = 1 - chi2.cdf(chi2_val, df)
    if chi2_val > chi2_crit:
        print(
            'Reject null hypothesis: Data is not MCAR (p-value={:.4f}, chi-square={:.4f})'.format(p_val, chi2_val)
        )
    else:
        print(
            'Do not reject null hypothesis: Data is MCAR (p-value={:.4f}, chi-square={:.4f})'.format(p_val, chi2_val)
        )
```

```python
little_mcar_test(data, alpha=0.05)
```

### `Résultats du test de Little pour l'hypothèse MCAR`

Le test de Little a été effectué pour vérifier si les données sont manquantes complètement au hasard (MCAR). Les résultats montrent que l'hypothèse nulle (les données sont MCAR) est rejetée, ce qui indique que les données ne sont **pas** manquantes complètement au hasard.

- **p-value** = 0.0000 (Une p-value très faible, bien inférieure à 0.05, ce qui signifie que l'hypothèse nulle est rejetée)
- **Chi-square** = 18824.1312 (Valeur du test chi carré)

Ces résultats suggèrent que les données manquantes ne sont probablement pas dues à un hasard pur, mais qu'elles pourraient être influencées par un autre facteur ou mécanisme spécifique. Cela signifie qu'il est nécessaire d'explorer davantage les causes des données manquantes avant de choisir une méthode d'imputation appropriée.

### `Méthode d'imputation bayésienne`

Étant donné que les données ne semblent pas être manquantes complètement au hasard, une approche adaptée pour imputer les valeurs manquantes pourrait être l'imputation bayésienne. Cette méthode permet de prendre en compte l'incertitude associée aux données manquantes en utilisant un modèle probabiliste.
